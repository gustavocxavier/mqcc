## Os dados

No laboratório anterior, "Distribuições amostrais", examinamos os dados da população de casas de Ames, Iowa. Vamos começar carregando esse conjunto de dados.

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import io
import requests

df_url = 'https://raw.githubusercontent.com/akmand/datasets/master/openintro/ames.csv'
url_content = requests.get(df_url, verify=False).content
ames = pd.read_csv(io.StringIO(url_content.decode('utf-8')))
```

Neste laboratório, começaremos com uma amostra aleatória simples de tamanho 60 extraída da população. Especificamente, esta é uma amostra aleatória simples de tamanho 60. Observe que o conjunto de dados contém informações sobre muitas variáveis habitacionais, mas, na primeira parte do laboratório, vamos nos concentrar na área da casa, representada pela variável `Gr.Liv.Area`.

```python
population = ames['Gr.Liv.Area']
samp = population.sample(60)
```

## Intervalos de confiança

Uma das formas mais comuns de descrever o valor típico ou central de uma distribuição é usar a média. Neste caso, podemos calcular a média da amostra usando:

```python
sample_mean = samp.mean()
```

Volte por um momento à pergunta que motivou este laboratório: com base nesta amostra, o que podemos inferir sobre a população? Considerando apenas esta única amostra, a melhor estimativa da área habitável média das casas vendidas em Ames seria a média amostral, geralmente denotada por $\bar{x}$ (aqui, nós a chamamos de `sample_mean`). Ela fornece uma boa *estimativa pontual*, mas também seria útil comunicar o grau de incerteza dessa estimativa. Isso pode ser representado por meio de um *intervalo de confiança*.

Podemos calcular um intervalo de confiança de 95% para uma média amostral somando e subtraindo 1.96 erros-padrão da estimativa pontual (consulte a Seção 4.2.3 caso esta fórmula não lhe seja familiar).

```python
import numpy as np

se = np.std(samp)/np.sqrt(60)
lower = sample_mean - (1.96 * se)
upper = sample_mean + (1.96 * se)
print(lower, upper)
```

Acabamos de fazer uma inferência importante: embora não saibamos como é a população completa, temos 95% de confiança de que a verdadeira área média das casas de Ames está entre os valores `lower` e `upper`. Algumas condições precisam ser atendidas para que esse intervalo seja válido.

Usando Python, vamos recriar muitas amostras para compreender melhor como as médias amostrais e os intervalos de confiança variam de uma amostra para outra.

Este é o roteiro geral:
* Obtenha uma amostra aleatória.
* Calcule e armazene a média e o desvio-padrão da amostra.
* Repita as etapas (1) e (2) 50 vezes.
* Use essas estatísticas armazenadas para calcular vários intervalos de confiança.

Antes de fazer tudo isso, precisamos criar arrays vazios nos quais possamos armazenar as médias e os desvios-padrão calculados para cada amostra (**DICA**: `numpy.empty()`). Aproveitaremos também para armazenar em `n` o tamanho de amostra desejado.

```python
samp_mean = np.empty(50)
samp_sd = np.empty(50)
n = 60
```

Agora estamos prontos para o laço em que calcularemos as médias e os desvios-padrão de 50 amostras aleatórias.

```python
for i in range(50):
    samp = population.sample(n) # obtém uma amostra de tamanho n = 60 da população
    samp_mean[i] = samp.mean() # armazena a média amostral no i-ésimo elemento de samp_mean
    samp_sd[i] = np.std(samp) # armazena o desvio-padrão amostral no i-ésimo elemento de samp_sd
```

Por fim, construímos os intervalos de confiança.

```python
se_array = samp_sd/np.sqrt(n)
lower_array = samp_mean - (1.96 * se_array)
upper_array = samp_mean + (1.96 * se_array)
```

Os limites inferiores desses 50 intervalos de confiança estão armazenados em `lower_array`, e os limites superiores, em `upper_array`. Vamos visualizar o primeiro intervalo.

```python
print(lower_array[1], upper_array[1])
```
